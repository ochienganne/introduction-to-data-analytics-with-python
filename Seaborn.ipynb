```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('Car data saved.xlsx')
df.head()                   
```

```python
variables= ['Min Price (Lakh)','Max Price (Lakh)','Range (kmpl)','CC','Seats','Variants',	'Type','Ex-Showroom Price','RTO','Insurance','Other','Onroad Price']
```

```python
df.columns
```

```python
plt.figure(figsize=(15,12))
for i,var in enumerate(variables,start=1):
    plt.subplot(4,3,i)
    sns.scatterplot(x=df[var],y=df['Max Price (Lakh)'], color='red',alpha=0.7)
    plt.title(f'scatter plot of {var} vs Onroad Price')
    plt.xlabel(var)
    plt.ylabel('Target')
plt.tight_layout()
plt.show()
    
```

```python

plt.figure(figsize=(15,10))
for i,var in enumerate(variables,start=1):
    plt.subplot(4,3,i)
    sns.histplot(df[var], bins=20, kde = True ,color='gray')
    plt.title(f'Histogram of {var}') 
    plt.xlabel(var)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
df = pd.read_excel('Car data saved.xlsx')
df.head()                   
```

```python
sns.boxplot(data=df['Max Price (Lakh)'])
plt.show()
```

```python
#Create a boxplot of range by car type
sns.boxplot(data=df,x='Type', y='Range (kmpl)', hue='Type',palette=['Red','Gray'])
plt.title('Boxplot of Range by Car Type',color='red')
plt.show()
```

```python
selected_columns=['Min Price (Lakh)','Max Price (Lakh)','Range (kmpl)','CC','Seats','Variants','Ex-Showroom Price','RTO','Insurance','Other']
df_selected=df[selected_columns]
df_selected.head()
```

```python
corr= df_selected.corr()
corr
```

```python
sns.heatmap(corr,cmap='BuPu')
plt.title('Heatmap of Correlation Matrix')
plt.show()
```

```python
sns.heatmap(corr, annot=True,cmap='viridis')
plt.title('Heatmap of Correlation Matrix')
plt.show()
```

```python
#Vilin plot of Max price by car Type
sns.violinplot(data=df, x='Type', y='Max Price (Lakh)', hue='Type',palette=['Blue','Green'])
plt.title('violin plot of Maximum Price by Car Type')
plt.show()
```

```python
sns.violinplot(data=df, x='Type', y='Range (kmpl)', hue='Type',palette=['Blue','Green'])
plt.title('violin plot of Range by Car Type')
plt.show()
```

```python
selected_columns=['Min Price (Lakh)','Max Price (Lakh)','Range (kmpl)','CC','Seats','Variants','Ex-Showroom Price','RTO','Insurance','Other','Type']
df_selected=df[selected_columns]
df_selected.head()
```

```python
#pairplot for the dataset
sns.pairplot(df_selected,hue='Type')
plt.suptitle('pairplot of Car Dataset',y=1.02)
plt.show()
```
